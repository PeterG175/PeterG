## Section 1: Displacement as a function of time
We have already looked at displacement-time graphs where we were told the journey someone has made (example: someone going to the shops and back home), and then using that we draw a displacement-time graph. Now we will look at how to change those graphs into an equation.  

### Example 1
Imagine we have Bob running a race at $5 m/s$. If I asked you how far Bob has travelled after 5 seconds, you can simply do $5 \times 5 = 25 m$ because we know that $\text{Distance} = \text{Speed} \times \text{Time}$. If I asked you how far Bob has travelled after 100 seconds, you will do a similar calculation and work out $5 \times 100 = 500$. Now if I ask you the question, _"How far will Bob have travelled after 't' seconds?"_ you know that if I ask you to calculate how far Bob has travelled at a certain time, then all you need to do is multiply the number by 5. If I say that $x$ represents the distance, and $t$ represents the time, then we can write the equation $$ x(t) = 5t$$

Notice that I have used $x(t)$ on the left have side of the equation. You may not have seen this before, so be careful because this does NOT mean $x \times t$. When you have a variable (in this case the distance) which is calculated only by using another variable (in this case the time), then we write $x(t)$ to tell whoever is reading our work that $t$ is the only thing used to calculate $x$. If I also wanted to use how sloped the ground was (represented by $s$) to calculate how far Bob has travelled, then I would have $x(s,t)$ because the distance Bob travels depends on both how long he has been running for, and also how sloped the ground is (if it is really steep then he might run slower).

Anyway, now that we have our equation for the displacement $x(t)=5t$, we can plot a displacement-time graph of Bob.

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,100,101)
y = 5*x

plt.plot(x,y)
plt.grid(True)
plt.title("Graph 1: Displacement time graph of Bob")
plt.xlabel("Time (minutes)")
plt.ylabel("Displacement (kilometers)")
plt.show()
```

Using this graph you can calculate values of $x(t)$ without doing any Maths. All you need to do it pick your value of time, $t$, find that value of time on the x-axis, then go up until you hit the line, then go to the left until you get to the y-axis. The value on the y-axis will give you your value for $x(t)$. For example, if I wanted to work out the displacement after 40 minutes, you would do the following:

## Section 2: Non-linear displacement functions
The title of this section may look tricky, but it means that within this section we will look at displacement-time graphs which are not a straight line. For example, imagine you are stood at the top of a building and you have just dropped a ball off it. The ball will speed up as it falls to the ground (I am assuming it does not reach termial velocity). This means that a straight line for our displacement-time graph will not accurately represent the ball, because the ball does not stay at the same speed while it is falling. For our displacement-time graph, we want it to have a small gradient at the start with the gradient getting steeper towards the end. One possible graph could be:

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,10,11)
y = x**2

plt.plot(x,y)
plt.grid(True)
plt.title("Graph 3: Displacement time graph of a ball falling")
plt.xlabel("Time (seconds)")
plt.ylabel("Displacement (meters)")
plt.show()
```

I have chosen to use the function $x(t)=t^2$ for this graph. This curve accuratly represents the path of the ball, but it does make calculating the speed of the ball trickier. With a straight line graph, to calculate the speed we need to pick two points and work out the gradient between them. With a straight line graph the gradient will be the same everywhere so it doesn't matter which two points we pick. However, with a curve that has a changing gradient this will not work. We discussed last week that if we want to calculate the speed, we need to choose an interval around the point that we want which is really small. The smaller the interval then the more accurate the value of the speed. We can now use some clever Maths to calculate the speed over a really small interval so we can be really accurate.

### Example 1: How to calculate speed from a non-linear displacement time graph
Imagine I have a ball falling, which can be represented by the function $x(t)=t^2$, and I want to find the speed of the ball after 3 seconds. Your first instinct might be to do $x(3)=3^2=9$, and say that the ball has travelled $9m$ in $3s$ so the ball must be travelling at $3ms^{-1}$. However, this is not true, because this is the average speed of the ball between $0s$ and $3s$ NOT the exact speed of the ball at $3s$. To find our speed of the ball at $3s$ we need to create our interval. One of the points I'm going to use is $(3,9)$ because we have already calculated it. Then I'm going to imagine that we waiting a small amount of time after 3 seconds. I don't know how long this small amount of time is, so I'm going to call it $\Delta t$. This would mean our other point is $(3+ \Delta t, (3+ \Delta t)^2)$. This can be seen in the graph below:

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = x**2

delta_t = 3.0

# Fixed point
x1, y1 = 3, 9

# Moving point
x2 = 3 + delta_t
y2 = x2**2

# Secant line: extend slightly beyond the two points
x_line = np.array([min(x1, x2) - 0.3, max(x1, x2) + 0.3])
slope = (y2 - y1) / (x2 - x1)
y_line = y1 + slope * (x_line - x1)

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x, y, color='steelblue', linewidth=2)
ax.plot(x_line, y_line, color='tomato', linewidth=2,
        linestyle='--')

ax.scatter([x1, x2], [y1, y2], color='tomato', zorder=5, s=60)

ax.annotate(r'$(3,\ 9)$',
            xy=(x1, y1), xytext=(x1 + 0.2, y1 + 2),
            fontsize=11)
ax.annotate(rf'$(3+\Delta t,\ (3+\Delta t)^2)$',
            xy=(x2, y2), xytext=(x2 + 0.2, y2 + 2),
            fontsize=11)

ax.set_xlim(0, 10)
ax.set_ylim(0, 100)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (meters)', fontsize=12)
ax.set_title('Graph 4: Displacement time graph of a ball falling', fontsize=13)
ax.grid(True)

plt.tight_layout()
plt.show()
```

Now that we have our interval, we can work out the speed of the ball within this interval (which is the same as the gradient of the orange line). We will do this as follows:
$$\text{Gradient} = \text{Speed} = \dfrac{\text{Change in displacement}}{\text{Change in time}} = \dfrac{(3+ \Delta t )^2 - 9}{(3 + \Delta t) -3}$$

We can now simplify this fraction to get:
$$ \dfrac{(3+ \Delta t )^2 - 9}{(3 + \Delta t) -3} = \dfrac{9 + 6 \Delta t + (\Delta t)^2 -9}{3 + \Delta t - 3} = \dfrac{6 \Delta t + (\Delta t)^2}{\Delta t}= 6 +\Delta t$$

Our final step requires using a limit. Last week we said that using the smallest interval possible will give the most accurate value for the speed. The smallest interval will happen when 3 is as close to $3 +\Delta t$ as possible, and this will happen when $\Delta t$ gets close to zero. So we can take the limit as $\Delta t$ approaches zero:
$$ \lim_{\Delta t \rightarrow 0} 6 + \Delta t = 6 $$
The above limit is shown by the diagram below. Notice how the orange line best represents the gradient (or the speed) when $\Delta t$ is smaller.

```python
#| echo: false
#| output: asis
import numpy as np
import plotly.graph_objects as go

x = np.linspace(0, 10, 500)
y = x**2

x1, y1 = 3, 9

def secant_points(dt):
    x2 = 3 + dt
    y2 = x2**2
    slope = (y2 - y1) / (x2 - x1)
    x_line = np.array([min(x1, x2) - 0.3, max(x1, x2) + 0.3])
    y_line = y1 + slope * (x_line - x1)
    return x2, y2, x_line, y_line

delta_t_values = np.round(np.arange(0.1, 5.1, 0.1), 1)

frames = []
for dt in delta_t_values:
    x2, y2, x_line, y_line = secant_points(dt)
    frames.append(go.Frame(
        name=str(dt),
        data=[
            go.Scatter(x=x, y=y),
            go.Scatter(x=x_line, y=y_line),
            go.Scatter(
                x=[x1, x2],
                y=[y1, y2],
                text=["", "(3+\u0394t, (3+\u0394t)\u00b2)"]
            )
        ]
    ))

dt0 = 3.0
x2_0, y2_0, x_line_0, y_line_0 = secant_points(dt0)

fig = go.Figure(
    data=[
        go.Scatter(
            x=x, y=y,
            mode='lines',
            showlegend=False,
            line=dict(color='steelblue', width=2)
        ),
        go.Scatter(
            x=x_line_0, y=y_line_0,
            mode='lines',
            showlegend=False,
            line=dict(color='tomato', width=2, dash='dash')
        ),
        go.Scatter(
            x=[x1, x2_0],
            y=[y1, y2_0],
            mode='markers+text',
            showlegend=False,
            marker=dict(color='tomato', size=8),
            text=["", "(3+\u0394t, (3+\u0394t)\u00b2)"],
            textposition=["top right", "top right"],
            textfont=dict(size=11)
        )
    ],
    frames=frames
)

steps = []
for dt in delta_t_values:
    steps.append(dict(
        method='animate',
        label=str(dt),
        args=[[str(dt)], dict(
            mode='immediate',
            frame=dict(duration=0, redraw=True),
            transition=dict(duration=0)
        )]
    ))

fig.update_layout(
    title='Graph 5: Displacement time graph of a ball falling',
    xaxis=dict(title='Time (seconds)', range=[0, 10]),
    yaxis=dict(title='Displacement (meters)', range=[0, 100]),
    showlegend=False,
    sliders=[dict(
        currentvalue=dict(prefix='Δt = ', font=dict(size=13)),
        pad=dict(t=50),
        steps=steps,
        active=list(delta_t_values).index(dt0)
    )],
    margin=dict(t=60, b=100)
)
import plotly.io as pio
pio.renderers.default = "notebook_connected"
fig.show()
```

In conclusion, after 3 seconds the ball is travelling at exactly $6ms^{-1}$.

### Example 2: How to calculate speed from a non-linear displacement time graph
Imagine that I am an Alien who drops a ball off a building. I live on a planet with higher gravity than earth, so the ball falls to the ground quicker than on earth. The ball on this planet if governed by $x(t)=t^3$ as it falls to the ground. What is the speed of the ball after 2 seconds?

Like in the last example, it would be useful to draw a graph. We will have ouor first point as $x(2)=2^3=8m$ and our second point will be at $2 + \Delta t$:

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = x**3

delta_t = 2.0

# Fixed point
x1, y1 = 2, 8

# Moving point
x2 = 2 + delta_t
y2 = x2**3

# Secant line: extend slightly beyond the two points
x_line = np.array([min(x1, x2) - 0.3, max(x1, x2) + 0.3])
slope = (y2 - y1) / (x2 - x1)
y_line = y1 + slope * (x_line - x1)

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x, y, color='steelblue', linewidth=2)
ax.plot(x_line, y_line, color='tomato', linewidth=2,
        linestyle='--')

ax.scatter([x1, x2], [y1, y2], color='tomato', zorder=5, s=60)

ax.annotate(r'$(2,\ 8)$',
            xy=(x1, y1), xytext=(x1 + 0.2, y1 + 2),
            fontsize=11)
ax.annotate(rf'$(2+\Delta t,\ (2+\Delta t)^3)$',
            xy=(x2, y2), xytext=(x2 + 0.2, y2 + 2),
            fontsize=11)

ax.set_xlim(0, 8)
ax.set_ylim(0, 100)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (meters)', fontsize=12)
ax.set_title('Graph 6: Displacement time graph of an Alien ball falling', fontsize=13)
ax.grid(True)

plt.tight_layout()
plt.show()
```

From this graph, we can work out the gradient as: $$ \text{Gradient} = \text{Speed} = \dfrac{\text{Change in displacement}}{\text{Change in time}} = \dfrac{(2 + \Delta t)^3 - 8}{(\ldots) - 2}$$

Simplifying this fraction, we will get: $$\dfrac{(2 + \Delta t)^3 - 8}{(\ldots) - 2}= \dfrac{\ldots - 8}{2 + \Delta t - 2} = \dfrac{12 \Delta t + 6 \Delta t^2 + \Delta t^3}{\Delta t}= 12 + 6 \Delta t + \Delta t^2$$
Then just like the last example, we can take the limit as $\Delta t \rightarrow 0$:
$$ \lim_{\Delta t \rightarrow 0} (12 + 6 \Delta t + \Delta t^2) = 12 ms^{-1}$$

$12 ms^{-1}$ is the speed of the ball governed by the equation $x(t)=t^3$ after 2 seconds.

## Section 3: Generalising finding the speed

In the last section we looked at finding the speed at a specific point using an equation for displacement that we have. However, what if instead of trying to find the speed, we tried to find an _equation_ for speed? We can do this using the exact same approach as what we did in the previous question.

### Example 1: Finding an equation for speed
Imagine I have a ball falling according to the equation $x(t)=t^2$. I want to find an equation which will tell me the speed of the ball at a certain time $t$. We will call this equation $v(t)$. We will do this example using the following few steps:

1. Pick two points to use
2. Draw a graph with these two points on
3. Find the gradient of these two points
4. Take the limit as $\Delta t$ approaches zero

Our first step is to find two points. We want to find a point at any general time $t$, so our first point will be the point $(t,t^2)$ (notice that this is the same as the point $(t,x(t))$ because $x(t)=t^2$). Then we will take another point which is close to our first point. So our second point will be $(t + \Delta t, (t+ \Delta t)^2)$. Now that we have our two points, we can draw a graph.

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = x**2

delta_t = 3.0

# Fixed point
x1, y1 = 3, 9

# Moving point
x2 = 3 + delta_t
y2 = x2**2

# Secant line: extend slightly beyond the two points
x_line = np.array([min(x1, x2) - 0.3, max(x1, x2) + 0.3])
slope = (y2 - y1) / (x2 - x1)
y_line = y1 + slope * (x_line - x1)

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x, y, color='steelblue', linewidth=2)
ax.plot(x_line, y_line, color='tomato', linewidth=2,
        linestyle='--')

ax.scatter([x1, x2], [y1, y2], color='tomato', zorder=5, s=60)

ax.annotate(r'$(t,\ t^2)$',
            xy=(x1, y1), xytext=(x1 + 0.2, y1 + 2),
            fontsize=11)
ax.annotate(rf'$(t+\Delta t,\ (t+\Delta t)^2)$',
            xy=(x2, y2), xytext=(x2 + 0.2, y2 + 2),
            fontsize=11)

ax.set_xlim(0, 10)
ax.set_ylim(0, 100)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (meters)', fontsize=12)
ax.set_title('Graph 7: Displacement time graph of a ball falling', fontsize=13)
ax.grid(True)

plt.tight_layout()
plt.show()
```

Using this, we can work out the gradient between our two points: $$\text{Gradient} = \text{Speed} = \dfrac{\text{Change in displacement}}{\text{Change in time}}=\dfrac{(t+\Delta t)^2 - t^2}{(t + \Delta t) - t}$$

Simplifying this fraction we will get: 
$$\dfrac{(t+\Delta t)^2 - t^2}{(t + \Delta t) - t}= \dfrac{t^2 + 2 t \Delta t + \Delta t^2 - t^2}{t + \Delta t - t} = \dfrac{2 t \Delta t + \Delta t^2}{\Delta t}= 2 t + \Delta t$$

Lastly, we can take the limit as $\Delta t$ approaches zero:
$$ \lim_{\Delta t \rightarrow 0}(2t + \Delta t) = 2t$$
This gives us our final equation for speed, $v(t) = 2t$. We can check that it works because in our other example we saw that the speed of a ball following the equation $x(t)=t^2$ after 3 seconds was equal to $6ms^{-1}$. Checking that with our new equation we get:
$$ v(3) = 2 \times 3 = 6 ms^{-1}$$
What we get with our new equation matches what we got earlier, so it is correct.

### Example 2: Finding an equation for speed
This is a difficult topic, so we can look at another example. We can do the same thing for example 2 in section 2, and try to find the equation for speed where the equation for displacement is $x(t)=t^3$. We can follow the same four steps as before. Our first step was to find two points that we can use to find the gradient between. One of our points will be $(t,t^3)$ and our other point will be $(t + \Delta t,(t + \Delta t)^3)$. Then we can draw our graph:

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = x**3

delta_t = 2.0

# Fixed point
x1, y1 = 2, 8

# Moving point
x2 = 2 + delta_t
y2 = x2**3

# Secant line: extend slightly beyond the two points
x_line = np.array([min(x1, x2) - 0.3, max(x1, x2) + 0.3])
slope = (y2 - y1) / (x2 - x1)
y_line = y1 + slope * (x_line - x1)

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x, y, color='steelblue', linewidth=2)
ax.plot(x_line, y_line, color='tomato', linewidth=2,
        linestyle='--')

ax.scatter([x1, x2], [y1, y2], color='tomato', zorder=5, s=60)

ax.annotate(r'$(t,\ t^3)$',
            xy=(x1, y1), xytext=(x1 + 0.2, y1 + 2),
            fontsize=11)
ax.annotate(rf'$(t+\Delta t,\ (t+\Delta t)^3)$',
            xy=(x2, y2), xytext=(x2 + 0.2, y2 + 2),
            fontsize=11)

ax.set_xlim(0, 8)
ax.set_ylim(0, 100)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (meters)', fontsize=12)
ax.set_title('Graph 8: Displacement time graph of an Alien ball falling', fontsize=13)
ax.grid(True)

plt.tight_layout()
plt.show()
```

Our next step is to find the gradient of these two points: $$\text{Gradient} = \text{Speed} = \dfrac{\text{Change in displacement}}{\text{Change in time}} = \dfrac{(t + \Delta t)^3 - t^3}{(t + \Delta t) - t} $$

Simplifying this fraction we get:
$$\dfrac{(t + \Delta t)^3 - t^3}{(t + \Delta t) - t} = \dfrac{(t^3 +3 t^2 \Delta t + 3 t \Delta t^2 + \Delta t^3) - t^3}{t + \Delta t - t}= \dfrac{3 t^2 \Delta t + 3 t \Delta t^2 +\Delta t^3}{\Delta t}= 3t^2 + 3t \Delta t + \Delta t^2$$

Then taking the limit as $\Delta t$ approaches zero we get:
$$\lim_{\Delta t \rightarrow 0}(3t^2 + 3t \Delta t + \Delta t^2) = 3t^2$$

So, when the equation for displacement is $x(t)=t^3$, the equation for the speed will be $v(t)=3t^2$.

In section 2 in example 2 we saw that after 2 seconds the speed of the ball would be $12 ms^{-1}$. We can check this using our new formula and get: $$v(2)=3 \times 2^2= 3\times 4 = 12ms^{-1}$$
