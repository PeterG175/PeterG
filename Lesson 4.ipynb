## Section 1: Using integration to find an equation for displacement

If we start with an equation of speed, $v(t)=3t^2$, then you will be able to recognise this as the derivative of $x(t)=t^3$, the equation for displacement could also be $x(t)=t^3+3$ or $x(t)=t^3-5$. When we looked at differenciation, we talked about how constants disappear when we differenciate. This means that many different equations of displacement can give the same equation for speed. So when we try and integrate, how do we know which equation of displacement is the correct one?

```python
#| echo: false
#| label: fig-diff-2
#| fig-cap: "A diagram showing what method to use to find equations."
from IPython.display import Image, display
display(Image(filename='Differentiating example.png'))
```

The answer to this question is that we don't know! Using only the information that the equation for speed is $v(t)=3t^2$ it is impossible to find the exact equation for displacment. Because of this, we can say that integrating $v(t)=3t^2$ with respect to time gives us $x(t) = t^3 +c$ where $c$ represents a constant (that we don't know the value of).

To find out the value of $c$, we need more information. A useful bit of information would be the displacement of the object for just one value of time. This can be seen in the following example:

### Example 1
Integrate the equation $v(t)=3t^2$ to find the equation of displacement. The displacement after 3 seconds is 20 $m$.

To answer this we can first say that the equation for displacement will be in the form $x(t)=t^3+c$. Then from the other information in the question, we know that $x(3)=20$. We can now put $t=3$ into our equation for displacement: $$x(t)=t^3+c \implies x(3)=3^3+c \implies(3)=27+c$$ Because we know that $x(3)=20$, then we can say that: $$27+c=20$$ From this we can see that $c=-7$. Our final equation for displacement is $x(t)=t^3-7$

Now to introduce some formulas. From our lesson on differentiation, we saw that if we have an equation for displacement in the form $x(t)=t^n$ then the equation for velocity will be $v(t)=nt^{n-1}$. However, as integration is the opposite of differentiation, we need to find the inverse of this formula. If we think about this logically, what we are doing is moving the power down and then taking away one from the power.

```python
#| echo: false
#| label: fig-diff-3
#| fig-cap: "A diagram showing what the formula for differentiation does."
from IPython.display import Image, display
display(Image(filename='Diff to int explanation.png'))
```

To find the inverse of this process, we need to start from the right side and do the opposite of what the arrows are showing. The opposite of 'Taking one off the power' will be 'Adding one to the power' and the opposite of 'Multiplying by the power' will be 'Dividing by the power'. We can show this in the diagram below. Remember in this diagram we are going from the right side to the left side.

```python
#| echo: false
#| label: fig-diff-4
#| fig-cap: "A diagram showing what the formula for integration does."
from IPython.display import Image, display
display(Image(filename='Int to diff explanation.png'))
```

This gives us our equation for integration, however there is one more step! We must not forget to include the '$+c$'  because we need more information to find out the exact equation for displacement. So to summarise, when we have an equation for speed in the form $v(t)=t^n$, then the equation for displacement will be in the form $x(t)=\dfrac{t^{n+1}}{n+1}+c$.
