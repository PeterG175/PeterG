## Section 1: Calculating speed
For this section I have made up these numbers, this is more about the understanding than the Maths.
Our first question we are going to answer is _"How can we calculate the speed of something?"_. This might seem obvious, because if you wanted to know the speed of someone who ran a 100 meter race in 15 seconds, then you would do the following calculation: $$ Speed=\dfrac{Distance}{Time}=\dfrac{100}{15}=6.67ms^{-1}$$ This means that if Bob did run a 100 _m_ in 15 _s_, then his average speed would be 6.67 _m/s_. However, we do know that he will not have travelled the entire race at this speed. During the start of the race he will be accellerating to his top speed, and he will be at his top speed in the middle of the race once he has fully accellerated. This means during the start he will be travelling less than 6.67 _m/s_ and at the middle of the race he will be travelling quicker than 6.67 _m/s_. So how would we calculate the speed of the Bob once he is exactly 15 _m_ into the race? As we are dealing with distance, it makes sense to change the distance that we are measuring over. Instead of working out his speed for the entire race, we can work out his speed for the first 30 _m_, as this will give us a more accurate value of how fast he is travelling at 15 _m_. If it took Bob 8 _s_ to run the first 30 meters, then we will get: $$ Speed = \dfrac{Distance}{Time}=\dfrac{30}{8}=3.75ms^{-1} $$ As you can see, this 3.75 _m/s_ is slower than 6.67 _m/s_ which is exactly what we expected. We can also decrease the distance we measure on even more, for example we can measure the time it takes Bob to run between 10 _m_ and 20 _m_. Measuring the time it takes Bob to run between 10 _m_ and 20 _m_ will give a more accurate value for the speed Bob is running at when he is exactly 15 _m_ into the race. This can be seen in figure 1.

```python
#| echo: false
#| label: fig-race-example
#| fig-cap: "A diagram showing a 100m race track, with the 15m mark labelled."
from IPython.display import Image, display
display(Image(filename='Race example.png'))
```

Now we can take a step back and think about what is happening. As we decrease the distance that we measure over, then the value of the speed becomes more accurate. So we want to measure over the smallest distance possible in to find the most accurate value for the speed of the person. Now we can have a go at some practice questions.

## Section 2: Displacement time graphs
Note that for this section I am using displacement and velocity instead of distance and speed. For those of you who have not studied the differences between distance and displacement (or speed and velocity), then you can think of them as the same thing and you will still be able to follow along.
You should now have a good understanding of how speed is calculated, now we can look at what this means on a graph. Displacement time graphs are commonly used to solve problems using speed. Displacement is displayed on the y-axis of the graph and time is displayed on the x-axis of the graph. For example, if I walked into university, then sat through a lecture and then got the bus back home, my displacement time graph would look something like the following:

```python
#| echo: false

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,2,1001)
y = np.piecewise(x,
                 [x<0.5, (x>=0.5) & (x<1.5), (x>=1.5) & (x<5.5/3), x>=5.5/3],
                 [lambda x: 2*x, lambda x:1, lambda x: -3*x+5.5, lambda x: 0])
plt.plot(x,y)
plt.grid(True)
plt.title("Graph 1: Displacement Time graph of my trip into University")
plt.xlabel("Time (hours)")
plt.ylabel("Displacement (kilometers)")
plt.show()
```

From graph 1 we can see a few things. The y-axis tells us that the University is 1km away from my house, as the line becomes horizontal at the 1km mark. The x-axis is telling us that it has taken me 30 minutes (or 0.5 hours) to walk into University, then spent 1 hour at University, then it took me around 20 minutes to get home on the bus. For a displacement time graph, the gradient of the graph represents the velocity. This can be seen because the gradient of the graph when I take the bus is steeper than the gradient of the graph when I walk, which implies that I am moving quicker when I'm on the bus, and slower when I'm walking. We can also see that the gradient of the graph is zero for the duration of my lecture, which means that I am not moving while watching my lecture. The last comment I want to make about this graph is that it is not perfectly accurate. We can see that the gradients of each portion of the graph is a straight line, which implies that I am moving at a constant velocity. We know this is not true because in real life you may have to stop or speed up for certain reasons (i.e. while I'm walking I may have to stop at traffic lights or while I'm on the bus the bus might wait at a bus stop). From this graph, there are two key takeaways:

1. The gradient of a displacement time graph represents the velocity
2. The steeper the gradient, then the greater the velocity

## Section 3: What is a limit?
In Maths, a limit is used to say that if we have some number _x_ that is getting closer and closer to another number _y_ , then _x=y_ . For example, if we have some number _x_ , that keeps getting closer and closer to the number 3, then we will write: $$ \lim_{x \rightarrow 3} x = 3 $$ One way that I like to think of limits is by thinking of an ant on a number line. As you can see from figure 3, the ant is starting from the left and it will keep moving to the right until it reaches the number 3, then it will stop.

```python
#| echo: false
#| label: fig-ant-number-line
#| fig-cap: "An ant moving slowly along a number line until it reaches the number 3"
from IPython.display import Image, display
display(Image(filename='Ant on number line.png'))
```

For this example it does not matter if the ant starts at the number -1 and move to the right (like it has in the diagram), or if the ant starts at the number 6 and moves to the left, as it will always finish at the number 3. There can be times when the value of the limit does depend on whether your increasing or decreasing with your value of _x_ (if your ant is moving to the left or to the right). One of the practice questions I have included for this section shows this and I have included it to challenge you so do not worry if you struggle with it, as we will talk about it when we have the lesson.

All of that aside, most limits I will show you can simply be solved be a direct substitution. For example, if I have the following: $$ \lim_{x \rightarrow 2} x^2-4x+5 $$ Then all I need to do to solve this is substitute in _x=2_ , which will give me: $$ \lim_{x \rightarrow 2} x^2-4x+5 = (2)^2-4(2)+5=4-8+5=1$$
